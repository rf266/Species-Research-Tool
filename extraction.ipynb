```python
import pickle
import os 
import nest_asyncio

from llama_index.core.extractors import TitleExtractor, KeywordExtractor

```

```python
with open("nodes.pkl", "rb") as f: 
    nodes = pickle.load(f)
```

```python
from llama_index.llms.openrouter import OpenRouter
from llama_index.llms.fireworks import Fireworks
from llama_index.llms.cerebras import Cerebras
import os
from dotenv import load_dotenv

load_dotenv()
api1 = os.getenv("OR_API_KEY")
model = OpenRouter(model="qwen/qwen3.7-plus", api_key =api1 )
api2 = os.getenv("CEREBRAS_API_KEY")
model2 = Cerebras(model="gpt-oss-120b", api_key=api2)
```

```python
titles = TitleExtractor(nodes=1, llm=model2)
keywords_extract = KeywordExtractor(keywords=10, llm=model2)

from llama_index.core.ingestion import IngestionPipeline

pipeline = IngestionPipeline(transformations=[titles, keywords_extract])

extracted_nodes = pipeline.run(documents=nodes, in_place=True, show_progress = True)
```
